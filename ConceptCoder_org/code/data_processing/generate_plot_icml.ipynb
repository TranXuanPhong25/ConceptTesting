```python
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import matplotlib.colors as mcolors
import numpy as np
```

```python
def plot_group(x_sorted, y_sorted, pred, model_names_sorted, color, title, filename):
    plt.figure(figsize=(6, 4))
    plt.scatter(x_sorted, y_sorted, color=color, label=f'{title} ΔV vs ΔC', s=30)
    plt.plot(x_sorted, pred, color=color, linestyle='--')

    for i, name in enumerate(model_names_sorted):
        plt.annotate(name, (x_sorted[i], y_sorted[i]), textcoords="offset points", xytext=(0, 6), ha='center', fontsize=8)

    plt.xlabel('ΔC')
    plt.ylabel('ΔV')
    # plt.title(title)
    plt.legend()
    plt.tight_layout()
    plt.savefig(filename, dpi=300)
    plt.show()
```

```python
plot_group(vd_x_sorted, vd_y_sorted, vd_pred, vd_model_names_sorted, 'blue', 'VD', './data/result/vd_regression_general_dataset.png')
```
